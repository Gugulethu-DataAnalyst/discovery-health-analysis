# CLEANING DATA

#Checking all null values in all dataframes


```python
#1. Check null values sum of null values in each column.
members_df.isna().sum()
```

```python
#In the members dataframe we find that we are only finding null values in the two data column; gender and region
#By the finding the best way to elliminate null variables of the data to ensure cean and accurate insight we can 
#In this instance we will fill null values to 'unknown' for both columns(genders, and region)
members_df["gender"] = members_df["gender"].fillna("Uknown")
members_df["region"] = members_df["region"].fillna("Unkown")

members_df.isna().sum()
```

```python
members_df.head()
```

#claim_data set null value check


```python
#claim data null check andsum
claims_df.isna().sum()
```

```python
#we find null values in 'diagnosis_code' 
claims_df['diagnosis_code'] = claims_df['diagnosis_code'].fillna("Uknown")
claims_df.isna().sum()
```

```python
#Currently the amounts paid and amount claimed are in string data type - need to convert to numeric
claims_df['amount_paid'] = pd.to_numeric(claims_df['amount_paid'].replace(',','',regex=True))
claims_df['amount_claimed'] = pd.to_numeric(claims_df['amount_claimed'].replace(',','',regex=True))


#we find null values and 'amounts_paid' - filling with unknown is the best way to keep data integrity

claims_df.head(5)
```

```python
#we find null values and 'amounts_paid' - filling with unknown is the best way to keep data integrity

claims_df['amount_paid'] = claims_df['amount_paid'].fillna(claims_df['amount_paid'].mean())
claims_df.isna().sum()
```

```python
claims_df.head(10)
```

#checking null values in Vitality data table 

```python
#checkung sum of null values in each columns
vitality_df.isna().sum()
```

```python
#we find them in 'steps', and 'points_earned'
#since we are working with sumabe data now - the best thong to fill null values is to place the mean value in the missing value placements
#first check the mean
round(vitality_df["steps"].mean(), 0)#how do i round up to 0 decimal places
```

```python
#Now i am filling missing values with null values in the steps column
vitality_df["steps"] = vitality_df["steps"].fillna(round(vitality_df["steps"].mean(),0))#rounding up to a whole number
#filling the points earned - with this column i will fill using the median operation
vitality_df["points_earned"] = vitality_df["points_earned"].fillna(round(vitality_df["points_earned"].median(),0))
vitality_df.isna().sum()
```

# DataType cleaning

```python
members_df.select_dtypes(include='object').columns
```

```python
#to turn the data column to str, this will change the type from object(errors) to string values
members_df['member_id'] = members_df['member_id'].astype(str)

#Changing to categorical type on these column
members_df['gender'] = members_df['gender'].astype('category')
members_df['plan_type'] = members_df['plan_type'].astype('category')
members_df['region'] = members_df['region'].astype('category')

#converting join_date data to be a datetime data type
members_df['join_date'] = pd.to_datetime(members_df['join_date'], errors='coerce')
```

```python
members_df.dtypes
```

```python
#SPLITTING THE NAME COLUMN TO TWO seperate first  name and surname - 
#for better data visulaition in the following stages
members_df[["FirstName","Surname"]] = members_df["name"].str.strip().str.split(' ',n=1, expand=True)

members_df.head()
```

```python
#dropping the name column
members_df.drop("name", axis=1, inplace=True)

##Restructuring the data columns
members_df = members_df[["member_id","FirstName", "Surname","age","gender","plan_type",
                         "chronic_condition","region","join_date"]]
members_df.head(10)
```

```python
#standardization of variations
members_df['region'] = members_df['region'].str.replace('.','', regex=False).str.upper().str.strip()
members_df.head(25)
```

```python
region_map={
    'ECAPE':'EASTERN CAPE',
}
members_df['region'] = members_df['region'].replace(region_map)
members_df.head(25)
```
